```python
import sqlite3
import pandas as pd

# 1. Create a connection to a local SQLite database
conn = sqlite3.connect('instacart.db')

# 2. Load your CSV files into pandas DataFrames and write them to SQL tables
print("Loading products...")
pd.read_csv('products.csv').to_sql('products', conn, if_exists='replace', index=False)

print("Loading order_products_train...")
pd.read_csv('order_products__train.csv').to_sql('order_products', conn, if_exists='replace', index=False)

print("Loading orders...")
pd.read_csv('orders.csv').to_sql('orders', conn, if_exists='replace', index=False)

print("Loading aisles...")
pd.read_csv('aisles.csv').to_sql('aisles', conn, if_exists='replace', index=False)

print("Loading departments...")
pd.read_csv('departments.csv').to_sql('departments', conn, if_exists='replace', index=False)

print("Database setup complete! All tables are ready for SQL queries.")
```

```python
import sqlite3
import pandas as pd

# 1. Connect to your database
conn = sqlite3.connect('instacart.db')

# 2. Write the SQL query for department reorder rates
query = """
SELECT
    d.department,
    COUNT(op.product_id) AS total_orders,
    SUM(op.reordered) AS total_reorders,
    ROUND(CAST(SUM(op.reordered) AS FLOAT) / COUNT(op.product_id) * 100, 2) AS reorder_rate_pct
FROM order_products op
JOIN products p ON op.product_id = p.product_id
JOIN departments d ON p.department_id = d.department_id
GROUP BY d.department
ORDER BY reorder_rate_pct DESC;
"""

# 3. Execute and display the results
df_result = pd.read_sql_query(query, conn)
print("--- Department Reorder Rates ---")
display(df_result.head(10))

# Close connection
conn.close()
```

```python
import sqlite3
import pandas as pd

# 1. Connect to your database
conn = sqlite3.connect('instacart.db')

# 2. Write the SQL query for orders by hour of the day
query_hours = """
SELECT
    order_hour_of_day,
    COUNT(order_id) AS total_orders
FROM orders
GROUP BY order_hour_of_day
ORDER BY order_hour_of_day;
"""

# 3. Execute and display the results
df_hours = pd.read_sql_query(query_hours, conn)
print("--- Orders by Hour of Day ---")
display(df_hours)

# Close connection
conn.close()
```

```python
import sqlite3
import pandas as pd

# 1. Connect to your database
conn = sqlite3.connect('instacart.db')

# 2. Write the SQL query for items per order
query_basket = """
SELECT
    order_id,
    COUNT(product_id) AS items_in_basket
FROM order_products
GROUP BY order_id;
"""

# 3. Calculate average and display summary stats using pandas
df_basket = pd.read_sql_query(query_basket, conn)
print("--- Basket Size Summary ---")
print(f"Average items per order: {df_basket['items_in_basket'].mean():.2f}")
display(df_basket['items_in_basket'].describe())

# Close connection
conn.close()
```
